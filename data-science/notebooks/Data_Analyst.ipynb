## 📉 Análisis de Comportamiento por Variables Demográficas

```python
# Función para análisis por segmento demográfico
def analyze_demographic_segment(df, segment_var, metric_var='total_sales'):
    """
    Analiza una métrica por segmento demográfico
    """
    analysis = df.groupby(segment_var).agg({
        metric_var: ['count', 'mean', 'median', 'std', 'min', 'max'],
        'customer_id': 'count'
    }).round(2)

    # Calcular percentiles
    percentiles = df.groupby(segment_var)[metric_var].agg(
        Q1=lambda x: x.quantile(0.25),
        Q3=lambda x: x.quantile(0.75)
    ).round(2)

    analysis['Q1'] = percentiles['Q1']
    analysis['Q3'] = percentiles['Q3']

    return analysis

# Análisis por grupo de edad
print("\n📈 COMPORTAMIENTO POR GRUPO DE EDAD:")
age_analysis = df_demo.groupby('age_group').agg({
    'total_sales': ['mean', 'median', 'std'],
    'avg_purchase_value': 'mean',
    'purchase_frequency': 'mean',
    'total_transactions': 'mean',
    'customer_id': 'count'
}).round(2)

display(age_analysis)


# Análisis por nivel de ingresos
if 'income_bracket' in df_demo.columns:
    print("\n📈 COMPORTAMIENTO POR NIVEL DE INGRESOS:")
    income_analysis = df_demo.groupby('income_bracket').agg({
        'total_sales': ['mean', 'median', 'std'],
        'avg_purchase_value': 'mean',
        'purchase_frequency': 'mean',
        'customer_id': 'count'
    }).round(2)

    display(income_analysis)

# Análisis por nivel educativo
if 'education_level' in df_demo.columns:
    print("\n📈 COMPORTAMIENTO POR NIVEL EDUCATIVO:")
    edu_analysis = df_demo.groupby('education_level').agg({
        'total_sales': ['mean', 'median'],
        'avg_purchase_value': 'mean',
        'age': 'mean',
        'customer_id': 'count'
    }).sort_values(('total_sales', 'mean'), ascending=False).round(2)

    display(edu_analysis)
```

## 🔬 Análisis Estadístico

```python
print("\n🔬 ANÁLISIS ESTADÍSTICO:")

# ANOVA: Diferencias en ventas por nivel de ingresos
if 'income_bracket' in df_demo.columns:
    print("\n📊 PRUEBA ANOVA - Ventas por Nivel de Ingresos:")
    income_groups = [df_demo[df_demo['income_bracket'] == level]['total_sales']
                    for level in income_order if level in df_demo['income_bracket'].unique()]

    if len(income_groups) >= 2:
        f_stat, p_value = f_oneway(*income_groups)
        print(f"  F-statistic: {f_stat:.3f}")
        print(f"  p-value: {p_value:.4f}")

        if p_value < 0.05:
            print("  ✅ CONCLUSIÓN: Hay diferencias significativas en ventas entre niveles de ingresos (p < 0.05)")
            # Prueba post-hoc (Tukey)
            from statsmodels.stats.multicomp import pairwise_tukeyhsd
            tukey = pairwise_tukeyhsd(endog=df_demo['total_sales'],
                                     groups=df_demo['income_bracket'],
                                     alpha=0.05)
            print("\n  📋 RESULTADOS TUKEY (post-hoc):")
            print(tukey)
        else:
            print("  ❌ CONCLUSIÓN: No hay diferencias significativas en ventas entre niveles de ingresos")

# Correlación edad-ventas
print(f"\n📊 CORRELACIÓN EDAD-VENTAS:")
corr_age_sales = df_demo['age'].corr(df_demo['total_sales'])
print(f"  Coeficiente de correlación: {corr_age_sales:.3f}")

if abs(corr_age_sales) > 0.3:
    strength = "fuerte" if abs(corr_age_sales) > 0.5 else "moderada"
    direction = "positiva" if corr_age_sales > 0 else "negativa"
    print(f"  ✅ Existe una correlación {direction} {strength} entre edad y ventas")
elif abs(corr_age_sales) > 0.1:
    print(f"  ⚠️ Existe una correlación débil entre edad y ventas")
else:
    print(f"  ❌ No existe correlación significativa entre edad y ventas")
```

# 🛒 Issue 2: Segmentación de Clientes por Comportamiento de Compra



## 🔠 Preparación de Datos para Segmentación

```python
print("\n" + "="*80)
print("ISSUE 2: SEGMENTACIÓN DE CLIENTES POR COMPORTAMIENTO DE COMPRA")
print("="*80)

# Crear copia del dataframe para segmentación
df_seg = df.copy()

# Seleccionar variables para segmentación
segmentation_vars = [
    'total_sales',
    'avg_purchase_value',
    'purchase_frequency',
    'total_transactions',
    'total_items_purchased',
    'avg_discount_used',
    'online_purchases',
    'in_store_purchases'
]

# Verificar que las variables existan
segmentation_vars = [var for var in segmentation_vars if var in df_seg.columns]
print(f"\n📋 Variables seleccionadas para segmentación ({len(segmentation_vars)}):")
print(segmentation_vars)

# Preparar datos para clustering
X = df_seg[segmentation_vars].copy()

# Estandarizar variables
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f"\n📊 Datos preparados para clustering:")
print(f"• Forma de X_scaled: {X_scaled.shape}")
print(f"• Variables: {segmentation_vars}")
```

## 🕵️‍♂️ Análisis de Componentes Principales (PCA)

```python
# Aplicar PCA para reducción de dimensionalidad
pca = PCA(n_components=min(5, len(segmentation_vars)))
X_pca = pca.fit_transform(X_scaled)


print("\n📊 ANÁLISIS DE COMPONENTES PRINCIPALES (PCA):")
print(f"• Varianza explicada por componente: {pca.explained_variance_ratio_}")
print(f"• Varianza acumulada: {np.cumsum(pca.explained_variance_ratio_)}")

# Visualizar varianza explicada
fig_pca_var = px.bar(x=[f'PC{i+1}' for i in range(len(pca.explained_variance_ratio_))],
                     y=pca.explained_variance_ratio_,
                     title='Varianza Explicada por Componente Principal',
                     labels={'x': 'Componente Principal', 'y': 'Varianza Explicada'},
                     color=pca.explained_variance_ratio_,
                     color_continuous_scale='Purp')
fig_pca_var.update_layout(yaxis_tickformat='.0%')
fig_pca_var.show()

# Scree plot
fig_scree = px.line(x=range(1, len(pca.explained_variance_ratio_)+1),
                    y=np.cumsum(pca.explained_variance_ratio_),
                    markers=True,
                    title='Varianza Acumulada - Scree Plot',
                    labels={'x': 'Número de Componentes', 'y': 'Varianza Acumulada'})
fig_scree.update_layout(yaxis_tickformat='.0%')
fig_scree.add_hline(y=0.8, line_dash="dash", line_color="red",
                    annotation_text="80% varianza", annotation_position="bottom right")
fig_scree.show()

```

## 💠 Determinación del Número Óptimo de Clusters

```python
# Método del codo y silhouette para determinar k óptimo
print("\n🔍 DETERMINANDO NÚMERO ÓPTIMO DE CLUSTERS...")

# Probar diferentes valores de k
k_range = range(2, 11)
inertia = []
silhouette_scores = []
calinski_scores = []

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)

    inertia.append(kmeans.inertia_)

    if len(X_scaled) > k:
        silhouette_scores.append(silhouette_score(X_scaled, kmeans.labels_))
        calinski_scores.append(calinski_harabasz_score(X_scaled, kmeans.labels_))

# Método del codo
fig_elbow = px.line(x=list(k_range), y=inertia, markers=True,
                    title='Método del Codo para K-Means',
                    labels={'x': 'Número de Clusters (k)', 'y': 'Inercia'})
fig_elbow.show()

# Método silhouette
fig_silhouette = px.line(x=list(k_range), y=silhouette_scores, markers=True,
                         title='Puntaje Silhouette por Número de Clusters',
                         labels={'x': 'Número de Clusters (k)', 'y': 'Puntaje Silhouette'})
fig_silhouette.show()

# Método Calinski-Harabasz
fig_calinski = px.line(x=list(k_range), y=calinski_scores, markers=True,
                       title='Puntaje Calinski-Harabasz por Número de Clusters',
                       labels={'x': 'Número de Clusters (k)', 'y': 'Puntaje Calinski-Harabasz'})
fig_calinski.show()
```

## 🖧 Aplicación de K-Means Clustering

```python
# Aplicar K-Means con k óptimo
optimal_k = 3
print(f"\n🎯 APLICANDO K-MEANS CON k = {optimal_k}")

kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=20)
df_seg['cluster'] = kmeans.fit_predict(X_scaled)

print(f"✅ Clustering completado. Distribución de clusters:")
cluster_dist = df_seg['cluster'].value_counts().sort_index()
for cluster, count in cluster_dist.items():
    percentage = (count / len(df_seg)) * 100
    print(f"  Cluster {cluster}: {count} clientes ({percentage:.1f}%)")
```

## 📚 Análisis de Características por Cluster

```python
# Analizar características de cada cluster
print("\n📊 CARACTERÍSTICAS POR CLUSTER:")

# Estadísticas por cluster
cluster_stats = df_seg.groupby('cluster')[segmentation_vars].agg(['mean', 'std', 'count']).round(2)

# Mostrar estadísticas
for cluster in sorted(df_seg['cluster'].unique()):
    print(f"\n🔵 CLUSTER {cluster}:")
    cluster_data = df_seg[df_seg['cluster'] == cluster]

    print(f"  • Tamaño: {len(cluster_data)} clientes ({(len(cluster_data)/len(df_seg))*100:.1f}%)")
    print(f"  • Ventas promedio: ${cluster_data['total_sales'].mean():,.0f}")
    print(f"  • Valor compra promedio: ${cluster_data['avg_purchase_value'].mean():,.0f}")
    print(f"  • Frecuencia compra: {cluster_data['purchase_frequency'].mean():.1f}")
    print(f"  • Descuento promedio usado: {cluster_data['avg_discount_used'].mean():.1%}")

    if 'online_purchases' in cluster_data.columns and 'in_store_purchases' in cluster_data.columns:
        online_ratio = cluster_data['online_purchases'].sum() / (cluster_data['online_purchases'].sum() + cluster_data['in_store_purchases'].sum())
        print(f"  • Preferencia online: {online_ratio:.1%}")

# Visualización de clusters en 2D usando PCA
fig_clusters_2d = px.scatter(
    x=X_pca[:, 0], y=X_pca[:, 1],
    color=df_seg['cluster'].astype(str),
    title='Segmentación de Clientes - Visualización 2D (PCA)',
    labels={'x': f'PC1 ({pca.explained_variance_ratio_[0]:.1%})',
            'y': f'PC2 ({pca.explained_variance_ratio_[1]:.1%})',
            'color': 'Cluster'},
    hover_data={'customer_id': df_seg['customer_id']} if 'customer_id' in df_seg.columns else None
)
fig_clusters_2d.update_traces(marker=dict(size=8, opacity=0.7))
fig_clusters_2d.show()

# Visualización 3D
if X_pca.shape[1] >= 3:
    fig_clusters_3d = px.scatter_3d(
        x=X_pca[:, 0], y=X_pca[:, 1], z=X_pca[:, 2],
        color=df_seg['cluster'].astype(str),
        title='Segmentación de Clientes - Visualización 3D (PCA)',
        labels={'x': f'PC1 ({pca.explained_variance_ratio_[0]:.1%})',
                'y': f'PC2 ({pca.explained_variance_ratio_[1]:.1%})',
                'z': f'PC3 ({pca.explained_variance_ratio_[2]:.1%})',
                'color': 'Cluster'},
        opacity=0.7
    )
    fig_clusters_3d.update_traces(marker=dict(size=5))
    fig_clusters_3d.show()

# Heatmap de características por cluster
cluster_means = df_seg.groupby('cluster')[segmentation_vars].mean()

# Estandarizar para mejor visualización
from sklearn.preprocessing import MinMaxScaler
scaler_heatmap = MinMaxScaler()
cluster_means_scaled = pd.DataFrame(
    scaler_heatmap.fit_transform(cluster_means),
    index=cluster_means.index,
    columns=cluster_means.columns
)

fig_heatmap = px.imshow(cluster_means_scaled.T,
                        labels=dict(x="Cluster", y="Variable", color="Valor Normalizado"),
                        title='Características Promedio por Cluster (Normalizado)',
                        color_continuous_scale='Purp',
                        aspect="auto")
fig_heatmap.update_xaxes(side="top")
fig_heatmap.show()
```

## 🔎 Análisis Exploratorio de Churn

```python
# Comparar características entre clientes que se quedan vs. se van
print("\n📈 COMPARACIÓN ENTRE CLIENTES ACTIVOS VS. CHURN:")

# Función para comparar grupos
def compare_groups(df, group_var, metric_vars):
    """
    Compara métricas entre grupos (churn vs no churn)
    """
    comparison = pd.DataFrame()

    for metric in metric_vars:
        if metric in df.columns:
            group_0 = df[df[group_var] == 0][metric]
            group_1 = df[df[group_var] == 1][metric]

            comparison.loc[metric, 'Churn_Mean'] = group_1.mean()
            comparison.loc[metric, 'No_Churn_Mean'] = group_0.mean()
            comparison.loc[metric, 'Diff_Abs'] = abs(group_1.mean() - group_0.mean())
            comparison.loc[metric, 'Diff_%'] = ((group_1.mean() - group_0.mean()) / group_0.mean()) * 100

            # Test t para diferencia de medias
            if len(group_0) > 1 and len(group_1) > 1:
                t_stat, p_val = ttest_ind(group_0, group_1, equal_var=False)
                comparison.loc[metric, 't-stat'] = t_stat
                comparison.loc[metric, 'p-value'] = p_val
                comparison.loc[metric, 'Significativo'] = p_val < 0.05

    return comparison.round(4)

# Variables a comparar
churn_metrics = [
    'total_sales', 'avg_purchase_value', 'purchase_frequency',
    'total_transactions', 'total_items_purchased', 'avg_discount_used',
    'days_since_last_purchase', 'membership_years',
    'online_purchases', 'in_store_purchases'
]

churn_comparison = compare_groups(df_churn, 'churn', churn_metrics)
print("\n📋 COMPARACIÓN DE MÉTRICAS:")
display(churn_comparison[['Churn_Mean', 'No_Churn_Mean', 'Diff_%', 'p-value', 'Significativo']])

# Visualización de diferencias
print("\n🎨 VISUALIZANDO DIFERENCIAS ENTRE GRUPOS...")

# Crear gráficos de barras comparativos
metrics_to_plot = ['total_sales', 'avg_purchase_value', 'purchase_frequency',
                   'days_since_last_purchase', 'membership_years']

fig_comparison = go.Figure()

for metric in metrics_to_plot:
    if metric in df_churn.columns:
        churn_mean = df_churn[df_churn['churn'] == 1][metric].mean()
        no_churn_mean = df_churn[df_churn['churn'] == 0][metric].mean()

        fig_comparison.add_trace(go.Bar(
            name=f'Churn ({metric})',
            x=['Churn', 'No Churn'],
            y=[churn_mean, no_churn_mean],
            visible=False
        ))

# Hacer visible solo el primero
fig_comparison.data[0].visible = True

# Crear botones
buttons = []
for i, metric in enumerate(metrics_to_plot):
    visibility = [False] * len(metrics_to_plot)
    visibility[i] = True

    buttons.append(
        dict(
            label=metric.replace('_', ' ').title(),
            method="update",
            args=[{"visible": visibility},
                  {"title": f"Comparación: {metric.replace('_', ' ').title()}"}]
        )
    )

fig_comparison.update_layout(
    title=f"Comparación: {metrics_to_plot[0].replace('_', ' ').title()}",
    updatemenus=[dict(
        active=0,
        buttons=buttons,
        x=1.05,
        xanchor="left",
        y=1.1,
        yanchor="top"
    )],
    showlegend=True
)

fig_comparison.show()


# Boxplots por categoría de churn
fig_boxplots = make_subplots(
    rows=2, cols=3,
    subplot_titles=[m.replace('_', ' ').title() for m in metrics_to_plot],
    vertical_spacing=0.1,
    horizontal_spacing=0.1
)

for i, metric in enumerate(metrics_to_plot[:6]):
    row = i // 3 + 1
    col = i % 3 + 1

    # Datos para churn
    churn_data = df_churn[df_churn['churn'] == 1][metric].dropna()
    no_churn_data = df_churn[df_churn['churn'] == 0][metric].dropna()

    fig_boxplots.add_trace(
        go.Box(y=churn_data, name='Churn', marker_color='red', boxmean=True),
        row=row, col=col
    )

    fig_boxplots.add_trace(
        go.Box(y=no_churn_data, name='No Churn', marker_color='blue', boxmean=True),
        row=row, col=col
    )

fig_boxplots.update_layout(
    height=600,
    title_text="Distribución de Métricas por Estado de Churn",
    showlegend=False
)
fig_boxplots.show()
```

## 🙋‍♂️ Análisis de Churn por Variables Demográficas

```python
print("\n👥 ANÁLISIS DE CHURN POR VARIABLES DEMOGRÁFICAS:")

# Función para analizar churn por variable categórica
def analyze_churn_by_category(df, category_var):
    """
    Analiza tasa de churn por categoría
    """
    if category_var not in df.columns:
        return None

    churn_by_cat = df.groupby(category_var).agg({
        'churn': ['count', 'mean', 'sum']
    }).round(4)

    churn_by_cat.columns = ['total_clientes', 'tasa_churn', 'churn_count']
    churn_by_cat['tasa_churn_%'] = churn_by_cat['tasa_churn'] * 100
    churn_by_cat['%_total_churn'] = (churn_by_cat['churn_count'] / churn_by_cat['churn_count'].sum()) * 100

    # Ordenar por tasa de churn
    churn_by_cat = churn_by_cat.sort_values('tasa_churn_%', ascending=False)

    return churn_by_cat

# Analizar por diferentes variables categóricas
demographic_vars_churn = ['gender', 'income_bracket', 'education_level',
                         'occupation', 'marital_status', 'loyalty_program']

for var in demographic_vars_churn:
    if var in df_churn.columns:
        print(f"\n📊 CHURN POR {var.upper()}:")
        analysis = analyze_churn_by_category(df_churn, var)
        if analysis is not None:
            display(analysis.head(10))

# Visualización de churn por variables categóricas
print("\n🎨 VISUALIZACIÓN DE CHURN POR CATEGORÍAS...")

top_cat_vars = []
for var in demographic_vars_churn:
    if var in df_churn.columns and df_churn[var].nunique() <= 10:
        top_cat_vars.append(var)

# Configuración de la rejilla
cols = 2
rows = math.ceil(len(top_cat_vars) / cols)

fig_cat_churn = make_subplots(
    rows=rows,
    cols=cols,
    subplot_titles=[f"Churn por {var.replace('_', ' ').title()}" for var in top_cat_vars],
    vertical_spacing=0.1,
    horizontal_spacing=0.1
)

# Iteración para añadir gráficos
for i, var in enumerate(top_cat_vars):
    churn_by_cat = analyze_churn_by_category(df_churn, var)

    if churn_by_cat is not None:
        # Calcular posición en la rejilla
        row = (i // cols) + 1
        col = (i % cols) + 1

        fig_cat_churn.add_trace(
            go.Bar(
                x=churn_by_cat.index.astype(str),
                y=churn_by_cat['tasa_churn_%'],
                name=var,
                marker_color='#5d4ac7',
                text=churn_by_cat['tasa_churn_%'].round(1).astype(str) + '%',
                textposition='outside',
                hovertemplate="<b>%{x}</b><br>Tasa: %{y:.1f}%<extra></extra>"
            ),
            row=row, col=col
        )

# Optimización del Layout
fig_cat_churn.update_layout(
    height=400 * rows,
    width=1000,
    title_text="<b>Análisis de Churn por Variables Categóricas</b>",
    title_x=0.5,
    template='plotly_white',
    showlegend=False,
    margin=dict(t=100, b=50, l=50, r=50)
)

# Ajustar ejes Y
fig_cat_churn.update_yaxes(title_text="Tasa (%)", ticksuffix="%", range=[0, 100])

# Rotar etiquetas de X
fig_cat_churn.update_xaxes(tickangle=45)

fig_cat_churn.show()
```

## 🛟 Análisis de Supervivencia (Kaplan-Meier)

```python
# Análisis de supervivencia
print("\n⏳ ANÁLISIS DE SUPERVIVENCIA:")

# Verificar información de membresía
if 'membership_years' in df_churn.columns:
    print("📊 CURVA DE SUPERVIVENCIA POR AÑOS DE MEMBRESÍA:")

    # Crear curva de supervivencia simple
    survival_data = []

    for years in sorted(df_churn['membership_years'].unique()):
        total_at_risk = len(df_churn[df_churn['membership_years'] >= years])
        churned_at_year = len(df_churn[(df_churn['membership_years'] == years) & (df_churn['churn'] == 1)])

        if total_at_risk > 0:
            survival_rate = 1 - (churned_at_year / total_at_risk)
            survival_data.append({
                'years': years,
                'at_risk': total_at_risk,
                'churned': churned_at_year,
                'survival_rate': survival_rate
            })

    survival_df = pd.DataFrame(survival_data)

    # Gráfico de curva de supervivencia
    fig_survival = px.line(survival_df, x='years', y='survival_rate',
                          markers=True,
                          title='Curva de Supervivencia - Retención por Años de Membresía',
                          labels={'years': 'Años de Membresía', 'survival_rate': 'Tasa de Supervivencia'},
                          line_shape='hv')

    fig_survival.update_layout(yaxis_tickformat='.0%')
    fig_survival.add_hline(y=0.5, line_dash="dash", line_color="red",
                          annotation_text="50% retención", annotation_position="bottom right")
    fig_survival.show()

    print(f"📈 Tasa de retención después de 1 año: {survival_df[survival_df['years'] == 1]['survival_rate'].values[0]:.1%}")
    print(f"📈 Tasa de retención después de 5 años: {survival_df[survival_df['years'] == 5]['survival_rate'].values[0] if 5 in survival_df['years'].values else 'N/A'}")
else:
    print("ℹ️ No se encontró información de membresía para análisis de supervivencia")
```

## 🎁 Promociones y Fidelización

```python
print("\n🤝 PROMOCIONES Y FIDELIZACIÓN:")

# Analizar relación entre promociones y churn/lealtad
if 'churn' in df_promo.columns and 'promo_flag' in df_promo.columns:
    print(f"\n📊 RELACIÓN ENTRE PROMOCIONES Y CHURN:")

    churn_by_promo = df_promo.groupby('promo_flag').agg({
        'churn': 'mean',
        'customer_id': 'count'
    })
    churn_by_promo['churn_rate_%'] = churn_by_promo['churn'] * 100
    churn_by_promo['diff_vs_avg'] = churn_by_promo['churn'] - df_promo['churn'].mean()

    display(churn_by_promo.round(4))

    # Test de significancia
    from scipy.stats import chi2_contingency

    contingency_table = pd.crosstab(df_promo['promo_flag'], df_promo['churn'])
    chi2, p_value, dof, expected = chi2_contingency(contingency_table)

    print(f"\n🔬 PRUEBA CHI-CUADRADO:")
    print(f"  Chi2: {chi2:.3f}, p-value: {p_value:.4f}")
    if p_value < 0.05:
        print(f"  ✅ La relación entre uso de promociones y churn es estadísticamente significativa")
    else:
        print(f"  ❌ No hay evidencia de relación significativa entre uso de promociones y churn")

# Análisis por programa de lealtad
if 'loyalty_program' in df_promo.columns:
    print(f"\n🎯 IMPACTO DEL PROGRAMA DE LEALTAD:")

    loyalty_analysis = df_promo.groupby('loyalty_program').agg({
        'total_sales': 'mean',
        'avg_purchase_value': 'mean',
        'purchase_frequency': 'mean',
        'churn': 'mean' if 'churn' in df_promo.columns else None,
        'promo_flag': 'mean' if 'promo_flag' in df_promo.columns else None,
        'customer_id': 'count'
    }).round(4)

    display(loyalty_analysis)

    # Visualización
    fig_loyalty = px.bar(loyalty_analysis.reset_index(),
                        x='loyalty_program',
                        y='total_sales',
                        title='Ventas Promedio por Participación en Programa de Lealtad',
                        labels={'loyalty_program': 'Programa de Lealtad',
                               'total_sales': 'Ventas Promedio ($)'},
                        color='total_sales',
                        color_continuous_scale='Purp')
    fig_loyalty.show()
```

## 🔖 Segmentación por Sensibilidad a Promociones

```python
# Crear segmentos basados en uso de promociones y descuentos
if all(col in df_promo.columns for col in ['promo_flag', 'avg_discount_used']):
    # Crear segmentos combinados
    conditions = [
        (df_promo['promo_flag'] == 0),  # No usan promociones
        (df_promo['promo_flag'] == 1) & (df_promo['avg_discount_used'] <= df_promo['avg_discount_used'].quantile(0.5)),  # Usan poco descuento
        (df_promo['promo_flag'] == 1) & (df_promo['avg_discount_used'] > df_promo['avg_discount_used'].quantile(0.5))  # Usan mucho descuento
    ]

    segment_names = ['No sensibles', 'Poco sensibles', 'Muy sensibles']
    df_promo['promo_sensitivity'] = np.select(conditions, segment_names, default='Otros')

    print(f"\n📊 DISTRIBUCIÓN DE SENSIBILIDAD:")
    sensitivity_dist = df_promo['promo_sensitivity'].value_counts(normalize=True) * 100
    for segment, percentage in sensitivity_dist.items():
        print(f"  {segment}: {percentage:.1f}%")

    # Análisis por segmento
    print(f"\n📈 COMPORTAMIENTO POR SENSIBILIDAD A PROMOCIONES:")
    sensitivity_analysis = df_promo.groupby('promo_sensitivity').agg({
        'total_sales': 'mean',
        'avg_purchase_value': 'mean',
        'purchase_frequency': 'mean',
        'avg_discount_used': 'mean',
        'churn': 'mean' if 'churn' in df_promo.columns else None,
        'customer_id': 'count'
    }).round(2)

    display(sensitivity_analysis)

    # Visualización
    fig_sensitivity = px.scatter(df_promo.sample(min(500, len(df_promo)), random_state=42),
                                x='avg_discount_used',
                                y='total_sales',
                                color='promo_sensitivity',
                                size='purchase_frequency',
                                hover_data=['customer_id', 'promotion_type'],
                                title='Segmentación por Sensibilidad a Promociones',
                                labels={'avg_discount_used': 'Descuento Promedio Usado',
                                       'total_sales': 'Ventas Totales ($)',
                                       'promo_sensitivity': 'Sensibilidad'},
                                color_discrete_sequence=['purple', 'mediumslateblue', 'gray'])
    fig_sensitivity.show()
```

# 🔗 Issue 5: Análisis de Correlaciones entre Variables Numéricas

## 🛠️ Preparación de Datos para Análisis de Correlaciones

```python
print("\n" + "="*80)
print("ISSUE #5: ANÁLISIS DE CORRELACIONES ENTRE VARIABLES NUMÉRICAS")
print("="*80)

# Crear copia para análisis de correlaciones
df_corr = df.copy()

# Seleccionar variables numéricas
numeric_cols = df_corr.select_dtypes(include=[np.number]).columns.tolist()

# Filtrar columnas que no son relevantes para correlación
exclude_cols = ['customer_id', 'churn', 'promo_flag', 'loyalty_program']
numeric_cols = [col for col in numeric_cols if col not in exclude_cols]

print(f"\n📋 VARIABLES NUMÉRICAS SELECCIONADAS ({len(numeric_cols)}):")
print(numeric_cols)
```

## 𝄜 Matriz de Correlación

```python
#Calcular matriz de correlación
print("\n📊 CALCULANDO MATRIZ DE CORRELACIÓN...")

correlation_matrix = df_corr[numeric_cols].corr()

# Mostrar matriz de correlación
print("\n🔍 MATRIZ DE CORRELACIÓN (primeras 10x10):")
display(correlation_matrix.iloc[:10, :10].round(3))

# Visualización de matriz de correlación completa
fig_corr_matrix = px.imshow(correlation_matrix,
                           labels=dict(color="Correlación"),
                           x=correlation_matrix.columns,
                           y=correlation_matrix.index,
                           title='Matriz de Correlación - Variables Numéricas',
                           color_continuous_scale='RdBu',
                           zmin=-1, zmax=1,
                           aspect="auto")
fig_corr_matrix.update_xaxes(side="top")
fig_corr_matrix.show()
```

## 💪 Análisis de Correlaciones Fuertes

```python
print("\n🔗 ANÁLISIS DE CORRELACIONES FUERTES:")

# Encontrar correlaciones fuertes (abs(corr) > 0.7)
strong_correlations = []

for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        corr_value = correlation_matrix.iloc[i, j]
        if abs(corr_value) > 0.7:
            var1 = correlation_matrix.columns[i]
            var2 = correlation_matrix.columns[j]
            strong_correlations.append({
                'Variable1': var1,
                'Variable2': var2,
                'Correlación': corr_value,
                'Tipo': 'Positiva' if corr_value > 0 else 'Negativa',
                'Fuerza': 'Muy Fuerte' if abs(corr_value) > 0.8 else 'Fuerte'
            })

strong_corr_df = pd.DataFrame(strong_correlations).sort_values('Correlación', ascending=False)

if len(strong_corr_df) > 0:
    print(f"\n📊 CORRELACIONES FUERTES ENCONTRADAS ({len(strong_corr_df)}):")
    display(strong_corr_df)
else:
    print("\nℹ️ No se encontraron correlaciones muy fuertes (|r| > 0.7)")

# Encontrar correlaciones moderadas (0.5 < |corr| <= 0.7)
moderate_correlations = []

for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        corr_value = correlation_matrix.iloc[i, j]
        if 0.5 < abs(corr_value) <= 0.7:
            var1 = correlation_matrix.columns[i]
            var2 = correlation_matrix.columns[j]
            moderate_correlations.append({
                'Variable1': var1,
                'Variable2': var2,
                'Correlación': corr_value,
                'Tipo': 'Positiva' if corr_value > 0 else 'Negativa'
            })

moderate_corr_df = pd.DataFrame(moderate_correlations).sort_values('Correlación', ascending=False)

if len(moderate_corr_df) > 0:
    print(f"\n📊 CORRELACIONES MODERADAS ENCONTRADAS ({len(moderate_corr_df)}):")
    # Mostrar solo las top 10
    display(moderate_corr_df.head(10))
else:
    print("\nℹ️ No se encontraron correlaciones moderadas (0.5 < |r| <= 0.7)")
```

## 🖇️ Visualización de Relaciones Clave

```python
print("\n🎨 VISUALIZANDO RELACIONES CLAVE...")

# Identificar las variables más correlacionadas con ventas
if 'total_sales' in correlation_matrix.columns:
    sales_correlations = correlation_matrix['total_sales'].sort_values(ascending=False)

    print(f"\n📊 VARIABLES MÁS CORRELACIONADAS CON VENTAS TOTALES:")
    top_correlations = sales_correlations[abs(sales_correlations) > 0.3].drop('total_sales')

    if len(top_correlations) > 0:
        display(top_correlations.round(3))

        # Gráfico de barras de correlaciones con ventas
        fig_sales_corr = px.bar(x=top_correlations.index, y=top_correlations.values,
                               title='Correlación con Ventas Totales',
                               labels={'x': 'Variable', 'y': 'Coeficiente de Correlación'},
                               color=top_correlations.values,
                               color_continuous_scale='RdYlGn',
                               range_color=[-1, 1])
        fig_sales_corr.update_layout(xaxis_tickangle=-45)
        fig_sales_corr.show()
    else:
        print("ℹ️ No hay variables con correlación significativa (>0.3) con ventas totales")

# %%
# Visualizar las relaciones más fuertes
if len(strong_corr_df) > 0:
    print("\n📈 VISUALIZACIÓN DE RELACIONES MÁS FUERTES:")

    # Tomar las 3 correlaciones más fuertes
    top_strong = strong_corr_df.head(3)

    for idx, row in top_strong.iterrows():
        var1 = row['Variable1']
        var2 = row['Variable2']

        fig_scatter = px.scatter(df_corr.sample(min(500, len(df_corr)), random_state=42),
                                x=var1, y=var2,
                                title=f'Relación entre {var1} y {var2} (r={row["Correlación"]:.3f})',
                                labels={var1: var1.replace('_', ' ').title(),
                                       var2: var2.replace('_', ' ').title()},
                                trendline='ols',
                                opacity=0.7)
        fig_scatter.show()

```

## 🔀 Análisis de Multicolinealidad

```python
# Calcular VIF (Variance Inflation Factor) para detectar multicolinealidad
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Preparar datos para VIF
X_vif = df_corr[numeric_cols].dropna()

# Calcular VIF para cada variable
vif_data = pd.DataFrame()
vif_data['Variable'] = X_vif.columns
vif_data['VIF'] = [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])]
vif_data = vif_data.sort_values('VIF', ascending=False)

print(f"\n📊 FACTOR DE INFLACIÓN DE VARIANZA (VIF):")
print("• VIF > 10 indica problemas serios de multicolinealidad")
print("• VIF > 5 indica posible multicolinealidad")
display(vif_data.round(2))

# Identificar variables con alto VIF
high_vif = vif_data[vif_data['VIF'] > 5]
if len(high_vif) > 0:
    print(f"\n⚠️ VARIABLES CON POSIBLE MULTICOLINEALIDAD (VIF > 5):")
    display(high_vif)

    print("\n💡 RECOMENDACIÓN: Considerar eliminar o combinar estas variables para modelos predictivos")
else:
    print("\n✅ No se detectaron problemas serios de multicolinealidad (VIF < 5 para todas las variables)")
```

## 🌀 Análisis de Componentes Principales (PCA) para Reducción

```python
print("\n📊 ANÁLISIS DE COMPONENTES PRINCIPALES (PCA) PARA REDUCCIÓN DE DIMENSIONALIDAD:")

# Estandarizar datos
scaler_pca = StandardScaler()
X_scaled_pca = scaler_pca.fit_transform(df_corr[numeric_cols].dropna())

# Aplicar PCA
pca_full = PCA()
X_pca_full = pca_full.fit_transform(X_scaled_pca)

# Análisis de varianza explicada
explained_variance = pca_full.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

print(f"\n📈 VARIANZA EXPLICADA POR COMPONENTES:")
for i, (var, cum_var) in enumerate(zip(explained_variance, cumulative_variance), 1):
    print(f"  PC{i}: {var:.1%} (Acumulado: {cum_var:.1%})")

# Determinar número óptimo de componentes
optimal_components = np.argmax(cumulative_variance >= 0.95) + 1
print(f"\n🎯 COMPONENTES PARA 95% DE VARIANZA: {optimal_components}")

# Visualizar scree plot
fig_scree_full = px.line(x=range(1, len(explained_variance)+1),
                         y=cumulative_variance,
                         markers=True,
                         title='Scree Plot - Varianza Acumulada',
                         labels={'x': 'Número de Componentes', 'y': 'Varianza Acumulada'},
                         line_shape='hv')
fig_scree_full.add_hline(y=0.95, line_dash="dash", line_color="red",
                        annotation_text="95% varianza", annotation_position="bottom right")
fig_scree_full.add_vline(x=optimal_components, line_dash="dash", line_color="green",
                        annotation_text=f"{optimal_components} componentes", annotation_position="top right")
fig_scree_full.update_layout(yaxis_tickformat='.0%')
fig_scree_full.show()


# Mostrar loadings de los primeros componentes
print("\n📊 LOADINGS DE LOS PRIMEROS 3 COMPONENTES PRINCIPALES:")

loadings = pd.DataFrame(
    pca_full.components_[:3].T,
    columns=[f'PC{i+1}' for i in range(3)],
    index=numeric_cols
)

# Mostrar las variables más importantes para cada PC
for pc in ['PC1', 'PC2', 'PC3']:
    print(f"\n🔍 VARIABLES MÁS IMPORTANTES PARA {pc}:")
    pc_loadings = loadings[pc].abs().sort_values(ascending=False).head(5)
    for var, loading in pc_loadings.items():
        sign = "+" if loadings.loc[var, pc] > 0 else "-"
        print(f"  {sign} {var}: {abs(loading):.3f}")
```
